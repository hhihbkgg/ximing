```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# 折线图
from matplotlib import rcParams #字体设置的函数    'font.sans-serif':函数中的字体设置方法
rcParams['font.sans-serif']=['SimHei'] #用来正常显示中文标签      win自带的字体'SimHei'

#1.创建画布设置大小  .figure(figsize=(10,6))       背景板大小(画布)
plt.figure(figsize=(10,6))

#2.数据
month=['1月','2月','3月','4月']
sales=[100,220,130,540]

#############        其它绘图只在绘制折线图这里稍作修改，其它都差不多，只是绘图方式略有不同      ##############
#3.绘制折线图   .plot(x轴数据,y轴数据)    添加图例：label("图例名")
plt.plot(month,sales,label="产品A",
         color='orange',linewidth=2,linestyle='--',marker='o') #linewidth线宽  marker='o'标记点形状为圆圈

#4.美化图表
#设置标题和字体大小以及颜色
plt.title('2026销量折线图',color='blue',fontsize=20)
#设置x轴(xlabel)和y轴的标签(ylabel)
plt.xlabel('月份',fontsize=10)
plt.ylabel('销量额(万元)',fontsize=10)
#控制图例位置    loc='位置'（在左上角显示）
plt.legend(loc='upper left')
#添加网格线   grid(axis='x' or 'y',alpha=0.1)   True为显示x和y网格线  axis='x'只显示x轴网格线   alpha透明度
plt.grid(True,alpha=0.1,color='black',linestyle='-') #"-"为实线,"--"为虚线
#设置刻度字体大小（x轴）
plt.xticks(fontsize=15, rotation=0)  #rotation=90为旋转90度显示
#设置y轴的起始范围 （默认最小值到最大值）
plt.ylim(0,600)

#对每一个点上显示数值 zip():将两个列表打包成一个个元组,然后进行遍历  plt.text(x,y,'显示的文字')
#x显示的横坐标位置(显示在月份上面）  显示的文字为：y的值   fontsize=10字体大小
#y+10为文字在y轴基础上往上偏移10个单位  va='bottom'文字底部对齐，ha='center'水平居中对齐
for x,y in zip(month,sales):
    plt.text(x,y+10,y,   fontsize=10,va='bottom',ha='center')

#全部显示出来
plt.show()
```

```python
#柱状图 bar()函数绘制柱状图

#1.创建画布设置大小  .figure(figsize=(10,6))       背景板大小(画布)
plt.figure(figsize=(10,6))

#2.数据
subject=['语文','数学','英语','物理','化学']
scores=[80,90,70,60,50]

#3.绘制柱状图   .bar(x轴数据,y轴数据)    添加图例：label("图例名")
plt.bar(subject,scores,label="小明",color='orange',width=0.5)

#4.美化图表
#设置标题和字体大小以及颜色
plt.title('学生成绩柱状图',color='blue',fontsize=20)
#设置x轴(xlabel)和y轴的标签(ylabel)
plt.xlabel('--科目--')
plt.ylabel('分数')
#控制图例位置    loc='位置'（在左上角显示）
plt.legend(loc='upper right')
#添加网格线   grid(axis='x' or 'y',alpha=0.1)    axis='y'只显示y轴网格线更美观   alpha透明度
plt.grid(axis="y",alpha=0.1,color='black',linestyle='-') #"-"为实线,"--"为虚线
#设置刻度字体大小（x轴）
plt.xticks(fontsize=10, rotation=0)  #rotation=90为旋转90度显示
#设置y轴的起始范围 （默认最小值到最大值）
plt.ylim(0,100)
#显示每个柱子上面的数值
for x,y in zip(subject,scores):
    plt.text(x,y+1,y,   fontsize=10,va='bottom',ha='center')

#全部显示出来
plt.show()
```

```python
#条形图 barh()函数绘制条形图(横向的柱状图)

#1.创建画布设置大小  .figure(figsize=(10,6))       背景板大小(画布)
plt.figure(figsize=(10,6))

#2.数据
countries=['America','China','Japan','Russia','UK']
gdp=[100,220,130,540,670]

#3.绘制条形图   .barh(y轴数据,x轴数据)    添加图例：label("图例名")
plt.barh(countries,gdp,label="国家Gdp",color='orange',height=0.5)

#4.美化图表
#设置标题和字体大小以及颜色
plt.title('2026国家GDP(万亿元）',color='blue',fontsize=20)
#设置x轴(xlabel)和y轴的标签(ylabel)
plt.xlabel('--Gdp--')
plt.ylabel('国家')
#控制图例位置    loc='位置'（在右下角显示）
plt.legend(loc='lower right')
#添加网格线   grid(axis='x' or 'y',alpha=0.1)    axis='x'只显示x轴网格线更美观   alpha透明度
plt.grid(axis="x",alpha=0.1,color='black',linestyle='-') #"-"为实线,"--"为虚线
#设置刻度字体大小（y轴）
plt.yticks(fontsize=10, rotation=0)  #rotation=90为旋转90度显示
#设置x轴的起始范围 （默认最小值到最大值）
plt.xlim(0,700)
#显示每个柱子上面的数值
for x,y in zip(gdp,countries):
    plt.text(x+1,y,x,   fontsize=10,va='center',ha='left')

plt.show()
```

```python
#饼图 pie()函数绘制饼图 看占比情况(2-6类最好）

#1.创建画布设置大小  .figure(figsize=(10,6))       背景板大小(画布)
plt.figure(figsize=(10,6))

#2.数据
activites=["学习","工作","娱乐","运动","其他"]
times=[4,3,5,2,1]
colors=['#66b3ff','#99ff99','#ffcc99','#ff9999','#99ff36'] #颜色列表
explode=[0.1,0,0,0,0] #explode列表，突出显示某一个部分  0.1:表示第一个部分偏离中心0.1个单位

#3.绘制饼图   times:按照时间做切分  lables=activites:标签名(代替了x轴和y轴) autopct:百分比显示
# "%.1f"保留一位小数  "%%"显示百分号,写两次转译%
#startangle:起始角度 (从90度开始画饼图，默认从0度开始画饼图)
#explode:突出显示某一个部分  (0.1,0,0,0,0)表示第一个部分突出显示，其它不突出
plt.pie(times,labels=activites,autopct='%.1f%%',
        startangle=90,explode=explode,
        colors=colors, #colors颜色列表
        shadow=True,#shadow:是否显示阴影效果

        #下面俩个参数是设置内部圆环的样式，不写也行
        wedgeprops={"width":0.5},  #wedgeprops:设置饼图内部圆环的宽度 (0.5为宽度的一半)
        pctdistance=0.7  #pctdistance:百分比显示的偏移距离
        )

#4.美化图表
#设置标题和字体大小以及颜色
plt.title('时间分布柱状图',color='black',fontsize=20)
#在圆环内部添加文字  text(x,y,显示的文字)   x=0 y=0为圆心位置
plt.text(0,0,"时间分布",fontsize=15,ha='center',va='center')

#全部显示出来
plt.show()
```

```python
#散点图 scatter() ：看许多个数据点之间的关系

#1.创建画布设置大小  .figure(figsize=(10,6))       背景板大小(画布)
plt.figure(figsize=(10,6))

#2.数据 y=2x+随机数 进行偏移
x=np.random.randint(1,1000,100) #生成100个随机数
y=2*x+np.random.randint(0,200,100) #生成100个随机数 y=2x+  随机数进行偏移

#3.绘制散点图   .scatter(x轴数据,y轴数据)
plt.scatter(x,y,alpha=0.5,color='orange',s=5,label="数据") #alpha透明度  s：点的大小

#4.美化图表
#设置标题和字体大小以及颜色
plt.title('y=2x',color='black',fontsize=20)
#设置x轴(xlabel)和y轴的标签(ylabel)
plt.xlabel('--x自变量--')
plt.ylabel('--y因变量--')
#添加网格线   grid(axis='x' or 'y',alpha=0.1)    alpha透明度
plt.grid(True,alpha=0.1,color='black',linestyle='-') #"-"为实线,"--"为虚线
#设置label图例
plt.legend(loc='upper left')

#画一条回归线 y=2x直线  [0,0]为起始点，[1000,2000]为终点
plt.plot([0,1000],[0,2100],label="回归线",color='blue',linewidth=2)
plt.legend(loc='upper left')

#全部显示出来
plt.show()
```

#### 一个画布上多个图的绘制  子图.图（）

```python
#1.创建画布设置大小 subplot(2,2,1)  2行2列的第1个图,共四个图
f1=plt.subplot(2,2,1)
f2=plt.subplot(2,2,2)
f3=plt.subplot(2,2,3)
f4=plt.subplot(2,2,4)

#2.数据
month=['1月','2月','3月','4月']
sales=[100,220,130,540]

#在子图上绘制折线图   子图.plot(x轴数据,y轴数据)
f1.plot(month,sales)
#在子图上绘制柱状图   子图.bar(x轴数据,y轴数据)
f2.bar(month,sales)
#在子图上绘制条形图   子图.barh(y轴数据,x轴数据)
f3.barh(month,sales)
#在子图上绘制饼图   子图.pie()
f4.pie(sales)

plt.show()
```

#### seaborn绘图库 相比于matplotlib 更简洁

```python
penguins=pd.read_csv('data/penguins.csv')
penguins.dropna(inplace=True)
penguins.head()
```

```python
#直方图+核密度估计图： data:数据集  x：横坐标列名 kde=True:显示核密度估计图
sns.histplot(data=penguins,x='bill_length_mm',kde=True)
plt.show()
```

```python
#核密度估计图：喙长度在x各值上的概率分布情况
sns.kdeplot(data=penguins,x='bill_length_mm')
plt.show()
```

```python
#箱形图：
sns.boxplot(x='species',y='bill_length_mm',data=penguins)
plt.show()
```

```python
#计数图
sns.countplot(x='species',data=penguins)
plt.show()
```

```python
#散点图 hue:分组依据的列名 ,不同颜色表示不同分组
sns.scatterplot(x='body_mass_g',y='flipper_length_mm',data=penguins,hue="sex")
plt.show()
```

#### 成对关系图：显示多个变量之间关系的图

```python
#按照species分组显示不同颜色的图
sns.pairplot(data=penguins,hue='species')
plt.show()
```
