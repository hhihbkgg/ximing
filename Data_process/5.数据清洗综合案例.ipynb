#### 企鹅案例
##### 1.导入库
##### 2.导入数据
##### 3.数据清洗
##### 4.数据特征的构造（拆开某列成两列，新加一些列和数据,转换类型等等）
##### 5.数据分析
##### 6.数据可视化

##### 分组聚合（按多个字段分组的单个聚合：实现单列的单个统计函数）:   group by(["分组字段1","分组字段2"])["聚合字段"].聚合函数()
##### 分组聚合（按多个字段分组的的多个聚合：实现多列的多个统计函数）:   group by(["分组字段1","分组字段2"]).agg({"聚合字段1":["统计  1","统计2"],...,"聚合字段2":["统计1","统计2"]})

```python
#1.导入库
import pandas as pd
import numpy as np

#2.导入数据
penguins=pd.read_csv('data/penguins.csv')
#检查部分数据以及其类型
display(penguins.head(10))
print(penguins.info())
print()

#3.数据清洗
#检查缺失值
print(penguins.isna().sum())
print()
#发现缺失值较少，直接原地删除
penguins.dropna(inplace=True)
print(penguins.isna().sum()) #缺失值都无了
print()

#4.数据特征的构造
#把sex转化成"category"分类类型
penguins["sex"]=penguins["sex"].astype("category")
print(penguins["sex"])
#添加一列长宽比
penguins["bill_ratio"]=round(penguins["bill_length_mm"]/penguins["bill_depth_mm"],2)
display(penguins.head(10))
print()

#5.数据分析
#把体重分为低中高3个等级 pd.cut(列，bins=n,labels=[分级名称列表])
labels=["low","medium","high"]
penguins["weight_level"]=pd.cut(penguins["body_mass_g"],bins=3,labels=labels)
display(penguins.head(10))
print()
#计算个等级的个数  列.value_counts():计算每个值的个数
print(penguins["weight_level"].value_counts())

#按照性别分组分析  统计的特别多时用 agg({"统计1"，"统计2"，"统计3"，...})合并起来 [sex]的【】可以不写，一个的分组字段
sex_group=penguins.groupby("sex").agg({
    "body_mass_g":["count","mean"] #统计企鹅个数和体重平均值
})
display(sex_group)
print()

#按照岛屿分组分析
island_group=penguins.groupby(["island"]).agg({
    "body_mass_g":["count","mean"], #统计企鹅个数和体重平均值
    "bill_ratio":["min","max"]  #计算长宽比的最小最大值
})
display(island_group)
print()
```

#### 睡眠质量案例

```python
#1.导入库
import pandas as pd
import numpy as np

#2.导入数据
sleep_quality=pd.read_csv('data/sleep.csv')
display(sleep_quality.head(10))
print()

#3.数据清洗
#检查缺失值
print(sleep_quality.isna().sum())
print()
#发现sleep_disorder缺失值较多，不可删除,进行填充   fillna(填充值)
sleep_quality.fillna("unknown",inplace=True)
print(sleep_quality.isna().sum()) #缺失值都无了
print()

#4.数据特征的构造
#genger转换为分类类型
sleep_quality["gender"]=sleep_quality["gender"].astype("category")
print(sleep_quality["gender"].dtypes)
print()

#输出职业的个数发现比较少    把职业occupation从str转为category分类类型
print(sleep_quality["occupation"].value_counts())
sleep_quality["occupation"]=sleep_quality["occupation"].astype("category")
print(sleep_quality["occupation"].dtypes)
print()

#血压值分成高低血压 split(列，分割符号) str:把blood_pressure当作字符串处理，然后分割符号为"/"  #expand=True:分割后变成两列
sleep_quality[["high_pressure","low_pressure"]]=sleep_quality["blood_pressure"].str.split("/",expand=True)
display(sleep_quality.head(10))
print()

#把睡眠质量分箱为“差，中，优”   pd.cut(列，bins=n,labels=[分级名称列表])
labels=["poor","medium","excellent"]
sleep_quality["sleep_level"]=pd.cut(sleep_quality["sleep_quality"],bins=3,labels=labels)
#把年龄分箱为“青年，中年，老年”
labels=["youth","middle_age","old"]
sleep_quality["age_level"]=pd.cut(sleep_quality["age"],bins=3,labels=labels)
display(sleep_quality.head(10))
print()

#5.数据分析
#输出BMI各等级的人数
print(sleep_quality["bmi_category"].value_counts())
print()

#根据不同的BMI分组，观察不同BMI人群的睡眠质量分布
#groupby("分组字段")[["聚合字段"]].value_counts()  单个分组字段，单个聚合字段
count=sleep_quality.groupby("bmi_category")["sleep_level"].value_counts()
display(count.to_frame("count"))
print()

#groupby(["分组字段"]).agg({"聚合字段1":["统计函数"],"聚合字段2":["统计函数"]})  多个分组字段，多个聚合字段
#根据年龄和BMI分组，计算不同年龄段不同BMI人群的睡眠时间和压力水平
BY_group=sleep_quality.groupby(["age_level","bmi_category"]).agg({
    "person_id":["count"], #计算人数
    "sleep_duration":["mean"],#计算睡眠时间平均值
    "stress_level":["mean"] #计算压力水平平均值
     })
#保留两位小数  列.round(2)
BY_group[["sleep_duration","stress_level"]]=BY_group[["sleep_duration","stress_level"]].round(2)
#给BY_group的"person_id"重命名
BY_group.rename(columns={"person_id":"person"},inplace=True)
display(BY_group)
```
