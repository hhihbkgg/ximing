#### 1.绘制气温趋势变化图
#### 2.绘制降水量直方图

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.sans-serif']=['SimHei'] #用来正常显示中文标签
```

```python
#导入数据
df=pd.read_csv('data/weather.csv')
display(df.head(10))
print(df.info()) #date列为str类型,需要转换为datetime类型.

#数据预处理
df['date']=pd.to_datetime(df['date']) #转换日期格式
print(df['date'].dtype) #转换成功

#绘制气温趋势变化图
plt.figure(figsize=(10,6))
df=df[df["date"].dt.year == 2015]#筛选出2015年数据,不画全部
plt.plot(df['date'],df['temp_max'],label="最高气温") #绘制x轴为日期,y轴为温度的折线图
plt.plot(df['date'],df['temp_min'],label="最低气温")
plt.title('2015年气温趋势变化图')
plt.xlabel('日期')
plt.ylabel('温度')
plt.legend()
plt.show()

#绘制绘制平均气温变化图
df['temp_avg']=(df['temp_max']+df['temp_min'])/2 #计算平均温度
display(df.head(5))
plt.figure(figsize=(10,6))
df=df[df["date"].dt.year == 2015]#筛选出2015年数据,不画全部
plt.plot(df['date'],df['temp_avg'],label="平均气温") #绘制x轴为日期,y轴为温度的折线图
plt.title('2015年平均气温趋势变化图')
plt.xlabel('日期')
plt.ylabel('温度')
plt.legend()
plt.show()
```

```python
#绘制降水量直方图
plt.figure(figsize=(10,6))
df=df[df["date"].dt.year == 2015]#筛选出2015年数据

#hist()函数绘制直方图,bins参数控制柱子的数量平均分成5份
plt.hist(df['precipitation'],bins=5,alpha=0.7)
plt.title('2015年平均气温趋势变化图')
plt.xlabel('降水量')
plt.ylabel('频数')
plt.show()
```
