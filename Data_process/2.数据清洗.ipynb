##### 1.缺失值NAN的处理 (np.nan,None,pd.NA)
###### np.nan:无限数或1/0这种错误(无法显示)
###### None:值为空，给它赋值为None，可以显示出来，但是不方便计算
###### pd.NA:没有填值也不赋值<NA>

```python
df1=pd.DataFrame(
    {
        "a":[1,2,3],
        "b":["a","b",None],
        "c":[np.nan,"c",pd.NA]
    }
)
display(df1)

#查看是否缺失值 df.isna() or df.isnull()

print(df1.isna())#df1.isnull()也可以
print()
print(df1.isna().sum())#统计缺失值:每一列缺失值的个数
print(df1.isna().sum(axis=1)) #统计缺失值:每一行缺失值的个数
```

```python
#剔除缺失值df.dropna()

#1.  df1.dropna() #默认剔除所有缺失值所在的行
df1.dropna(how="all") #how="all"某一行所有值都缺失时，剔除该行
#2.  df1.dropna(thresh=2) #某行至少有2个值不缺失时，保留该行

#3.df1.dropna(axis=1,how="all") #某一列所有值都缺失时，剔除该列
#4.df1.dropna(subset=["a"]) #剔除列"a"中缺失值的行
```

##### 2.填充缺失值
###### fillna:填充缺失值
###### replace:替换缺失值
##### fillna不会原地修改数据，需要赋值

```python
#weather中有缺失值
weather=pd.read_csv("data/weather_withna.csv")
display(weather)
print(weather.isna().sum(axis=0)) #axes=0表示按列统计缺失值个数

#填充缺失值 fillna()
weather=weather.fillna({"precipitation":20,'temp_max':10})
display(weather) #字典形式填充

wind_mean=weather["wind"].mean()
print(wind_mean)
weather=weather.fillna({"wind":round(wind_mean,3)})
display(weather) #填充平均值

#用周边值填充 df.ffill() :ffill(forward fill)向前填充
weather=weather.ffill()
display(weather)
```

#### 数据类型替换

```python
data={
    "name":["Alice","Alice","Bob","Tom","Bob","Alice"],
    "age":[20,20,22,20,21,25],
    "sex":["male","male","male","male","female","male"]
}
df=pd.DataFrame(data,columns=["name","age","sex"])
display(df)

#检测重复行 .duplicated()
print(df.duplicated()) #某行数据完全相同，返回True
display(df.drop_duplicates()) #剔除重复行:默认剔除完全相同的行

display(df.drop_duplicates(subset=["name"])) #剔除重复行:根据某一列剔除:剔除name列重复的行
#df.drop_duplicates(subset=["name"],keep="last") #保留最后一次出现的行
```

```python
#数据类型转换   列.astype(数据类型)
df=pd.read_csv("data/sleep.csv")
display(df)
print(df.dtypes)
#age的类型是int64,转换成小一点的数据类型int16。不占用太多内存

df["age"]=df["age"].astype(np.int16) #列的类型转换后赋值给列 原地修改数据
#display(df["age"].astype(np.int16)) #不原地修改数据,返回新的df表格，不推荐使用

#字符串处理比较慢，原来的genger为str类型，如：在比较str=="male"时太慢，转换成catagory类别型数据后比较快
df["gender"]=df["gender"].astype("category") #转换成类别型数据 catagory中有male,female
display(df)
print(df["age"].dtypes)
print(df["gender"].dtypes)
```

#### 数据变形()

```python
#pd.melt():数据变形,将宽表转为长表
#id_vars:不变列,value_vars:变列.value_name:变列的名字,var_name:取值列的名字
df=pd.DataFrame(
    {
        "name":["Alice","Bob"],
        "age":[20,21],
        "math":[90,80],
        "english":[80,90],
        "chinese":[80,90]
    },index=["A","B"]
)
display(df)
#不是原地修改数据，返回新的df表格
df1=pd.melt(df,id_vars=["name","age"],value_vars=["math","english","chinese"],value_name="score",var_name="subject")
df1.sort_values("name",inplace=True) #原地修改数据，不返回新的df表格，按照名字排序
display(df1)
```

```python
#分列 sleep.csv中有血压值，想要分开最高和最低血压值
sleep=pd.read_csv("data/sleep.csv")
sleep_part=sleep.head(10)
display(sleep_part)
print(sleep_part.dtypes["blood_pressure"]) #血压值124/70是字符串类型

#把高低血压分开，变成两列
#列.str:指对字符串处理(血压是str)   .split("/",expand=True):分割字符串，“/“是分隔符表示先分割成124,70, expand=True：变成两列
splt=sleep_part["blood_pressure"].str.split("/",expand=True) #分割字符串

sleep_part[["Maxblood_pressure","Minblood_pressure"]]=splt #赋值给两列 注意加两层[[]]

#类型转换,str转int16
sleep_part["Maxblood_pressure"]=sleep_part["Maxblood_pressure"].astype(np.int16)
sleep_part["Minblood_pressure"]=sleep_part["Minblood_pressure"].astype(np.int16)
print(sleep_part["Maxblood_pressure"].dtypes)
display(sleep_part)
```

##### 数据分箱：将连续数据分割成一个个区间  pd.cut(pd,bins,labels)
###### pd:数据列  bins:区间个数或区间范围列表 labels:每个区间的标签(名字)

```python
df=pd.read_csv("data/employees.csv")
df1=df.head(10)[["employee_id","salary"]]
display(df1)

#将薪资划分成2个区间，看每行的薪资落在哪个区间
#bins=整数2:    分成两个区间;  把salary.min~salary.max:薪资范围分成两个区间:4180-14100-2400
print(pd.cut(df1["salary"],bins=2))
print()
count=pd.cut(df1["salary"],bins=2).value_counts() #统计每个区间的人数
print(count)
```

```python
#用法2:bins=自定义区间列表 labels=区间名字
print(pd.cut(df1["salary"],bins=[0,10000,20000,30000],labels=["低","中","高"]))
df1["收入范围"]=pd.cut(df1["salary"],bins=[0,10000,20000,30000],labels=["低","中","高"])
display(df1)
```

##### pd.qcut(pd,q,labels):将数据分成几个等频区间(每个区间里的人数差不多)

```python
print(pd.qcut(df1["salary"],q=3,labels=["低","中","高"]))
#统计每个区间的人数,可以发现每个区间的人数差不多    等频区间
print(pd.qcut(df1["salary"],q=3,labels=["低","中","高"]).value_counts())
```

```python
#使用sleep.csv数据，将睡眠时长分成3个区间
df=pd.read_csv("data/sleep.csv")
df1=df.head(10)[["person_id","sleep_quality"]] #取前10行,只取id和total_sleep两列
display(df1)

#数值——>分箱实现类别——>统计
df1["睡眠质量"]=pd.cut(df1["sleep_quality"],bins=3,labels=["低","中","高"])
display(df1)
print(df1["睡眠质量"].value_counts())
print()

#.astype("category")转换成类别型数据,并统计的方法   字符串——>类别——>统计     对数值操作需要分箱转化为类别再统计
df["gender"]=df["gender"].astype("category") #转换成分类型数据 catagory:{"male","female"}
count=df["gender"].value_counts()
print(count)
print()

#都是catagory类型数据
print(df["gender"])
print(df1["睡眠质量"])
```
