#### 时间数据的处理
##### pd.Timestamp("2026-09-22 17:24)   (时间戳)时刻

```python
time=pd.Timestamp("2026-09-22 17:24")
print(time)
print(type(time))

#属性
print(time.year)
print(time.month)
print(time.day)
print(time.hour)
print(time.minute)
print(time.second)
print(time.quarter)
print()

#方法
print("是否是月底",time.is_month_end)
print("星期",time.day_name())
print("转化为天",time.to_period("D")) #转化为天,月M,年Y,周W,小时H,分钟T,秒S
print("转化为月",time.to_period("M")) #转化为月
```

```python
#字符串转换成时间戳类型(我们导入来的数据里的时间多是字符串类型)
time_str="2026-09-22 17:24"
print(pd.to_datetime(time_str))
print(type(pd.to_datetime(time_str)))
print()

#DataFame中的例子
df=pd.DataFrame(
    {
    "name":["Alice","Bob"],
    "生日":["1990-03-15","1988-05-20"]
}
)
print(df["生日"].dtype)
#将字符串转化为时间戳类型
df["生日"]=pd.to_datetime(df["生日"])
print(df["生日"].dtype)
print()

print(df["生日"].dt.year) #直接df["生日"].year会出错       df["生日"]是Series类型    里面的属性才是时间戳类型  dt:datetime
#单个时间戳才直接.year  对于属性为时间戳的Series要用.dt 把Series里面的时间戳类型提取出来

df["weekday"]=df["生日"].dt.day_name()
df["year"]=df["生日"].dt.year
display(df)
```

#### 分组聚合:   group by("分组字段")["聚合字段"].聚合函数()


```python
df=pd.read_csv("data/employees.csv")

#先进行数据清洗

#id有缺失值，将缺失值先处理，才能对ID进行类型转换和分组 (缺失值处理：填充，删除)
print(df["department_id"].isna().sum())
#只有一行缺失值，直接删除该行就行 (inplace=True 直接在原数据上修改)
df.dropna(subset=["department_id"],inplace=True) #只检查department_id列是否有缺失值，若有则删除该行
#将department_id由浮点型转化为整型
df["department_id"]=df["department_id"].astype("int16")
display(df)

#按照部门ID分组，计算每个部门的员工数量和平均薪资 group by
#("ID")["薪资"].count()  :按照ID分组，对[薪资]计数  .count()是聚合函数
count=df.groupby("department_id")["employee_id"].count()
mean=df.groupby("department_id")[["salary"]].mean()
#[[]]表示返回的时DataFrame类型  只[]表示返回的是Series类型
display(count.to_frame())
print()

#处理mean中salay的小数点后两位
mean["salary"]=round(mean["salary"],2)
mean.reset_index(inplace=True)#重置索引，便于观察
mean.sort_values("salary",ascending=False,inplace=True) #按照salary降序排列,inplace=True 直接在原数据上修改
display(mean)

```

```python
#多条件分组：根据不同部门(department_id)，不同岗位(job_id)的平均薪资
mean=df.groupby(["department_id","job_id"])[["salary"]].mean() #[[salay]]:返回DataFrame类型
mean=mean.reset_index()
mean["salary"]=round(mean["salary"],2)
mean.sort_values("salary",ascending=False,inplace=True)
display(mean)
```
